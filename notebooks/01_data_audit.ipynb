```python

# Hapus "Rp ", hapus koma, dan ubah menjadi integer
df['harga'] = (
    df['harga']
    .str.replace('Rp', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype(int)
)

# Ubah teks 'New' menjadi NaN (kosong), lalu ubah menjadi float
df['rating'] = pd.to_numeric(df['rating'].replace('New', np.nan), errors='coerce')

# 3. Cek hasilnya
print("--- INFO SETELAH DIBERSIHKAN ---")
df.info()

print("\n--- STATISTIK DESKRIPTIF ---")
display(df[['harga', 'rating', 'jumlah_review']].describe())
```

```python
# Ubah teks 'N/A' menjadi NaN (kosong), lalu ubah menjadi float
df['rating'] = pd.to_numeric(df['rating'].replace('N/A', np.nan), errors='coerce')
```

## Ringkasan Audit Data (Fase 1)

- **Dimensi Data**: Terdapat 572 baris dan 6 kolom.
- **Tipe Data**: 
  - `harga` awalnya berupa string (mengandung "Rp " dan koma) dan berhasil dibersihkan menjadi integer (`int64`).
  - `rating` awalnya berupa string (mengandung teks 'New') dan berhasil diubah menjadi desimal (`float64`).
- **Missing Values**:
  - `rating` memiliki 52 nilai kosong (sekitar 9%). Disebabkan oleh baris yang berupa hunian baru sehingga atribut di kolom rating berupa 'New' atau 'N/A'
  - Kolom lain tidak memiliki nilai kosong.
- **Distribusi Harga**:
  - Sangat *right-skewed* (condong ke kanan). Median harga adalah Rp 7.5 juta, namun rata-ratanya mencapai Rp 10.4 juta.
  - Terdapat harga ekstrem (outlier) hingga Rp 94 juta, dan terendah Rp 925 ribu. Transformasi logaritma (`log`) akan diperlukan pada fase pemodelan.
- **Distribusi Rating**:
  - Sangat terkonsentrasi di angka tinggi (rentang 4.29 hingga 5.0, dengan median 4.93). Standar deviasinya sangat kecil (0.11).
  - Hal ini menandakan bahwa variasi rating antar-listing sangat minim, sehingga prediksi harga berdasarkan rating murni mungkin akan memiliki korelasi yang lemah.
- **Lokasi & Wilayah**:
  - Kolom `wilayah` belum divalidasi. (Akan dicek normalisasinya di Fase 2).
